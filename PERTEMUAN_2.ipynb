C. Penyisipan di akhir

```python
class Node:
  def __init__ (self, dataval=None):
    self.dataval = dataval
    self.nextval = None

#create node
n1 = Node("Januari")
n2 = Node("Februari")

class LinkedList:
  def __init__(self):
    self.headval = None

# Print the linked list
  def listprint(self):
    printval = self.headval
    while printval is not None:
      print (printval.dataval)
      printval = printval.nextval

# Function to add newnode in the end
  def AddInEnd(self, newdata):
    NewNode = Node(newdata)
    if self.headval is None:
      self.headval = NewNode
      return
    last = self.headval
    while(last.nextval):
      last = last.nextval
    last.nextval=NewNode

Li = LinkedList()
Li.headval = n1

#Link first Node to second node
Li.headval.nextval = n2

#add node in the end
Li.AddInEnd("The Last")
Li.listprint()
```

# Menghapus node pada linked list

Hapus node dengan kata kunci

```python
class Node:
  def __init__(self, dataval=None):
    self.dataval = dataval
    self.nextval = None

class LinkedList:
  def __init__ (self):
    self.headval = None

# Print the linked list
  def listprint(self):
    printval = self.headval
    while printval is not None:
      print (printval.dataval)
      printval = printval.nextval

#Function to add node in the begining
  def AddBegining(self,newdata):
    NewNode = Node (newdata)
#Update the new nodes next val to existing node
    NewNode.nextval = self.headval
    self.headval = NewNode

#Function to remove node by key
  def RemoveNode(self, Removekey):
     HeadVal = self.headval
     if (HeadVal is not None):
        if (HeadVal.dataval == Removekey):
           self.headval = HeadVal.next
           HeadVal = None
           return
     while (HeadVal is not None):
        if HeadVal.dataval == Removekey:
           break
        prev = HeadVal
        HeadVal = HeadVal.nextval
     if (HeadVal == None):
        return
     prev.nextval = HeadVal.nextval
     HeadVal = None

#create linked list named Li
Li = LinkedList()

#add node
Li.AddBegining("April")
Li.AddBegining("Maret")
Li.AddBegining("Febuari")
print("Sebelum dilakukan remove")
Li.listprint()
print("================================================")

#remove node by key "Maret"
Li.RemoveNode("Maret")
print("Setelah dilakukan remove dengan kunci 'Maret'")
Li.listprint()
```

```python
class Node:
   def __init__(self,dataval=None):
      self.dataval = dataval
      self.nextval = None

class LinkedList:
  def __init__(self):
    self.headval = None

# Print the linked list
  def listprint(self):
    printval = self.headval
    while printval is not None:
       print (printval.dataval)
       printval = printval.nextval

# function to add newnode in the end
  def AddInEnd(self,newdata):
      NewNode = Node(newdata)
      if self.headval is None:
         self.headval = NewNode
         return
      last = self.headval
      while(last.nextval):
        last = last.nextval
      last.nextval=NewNode

# function to remove first node
  def RemoveEnd(self):
    last = self.headval
    while (last is not None):
      if last.nextval == None:
        break
      prev = last
      last = last.nextval
    if (last == None):
      return
    prev.nextval = last.nextval
    last = None

# create linked list named Li
Li = LinkedList()

# add node
Li.AddInEnd("Januari")
Li.AddInEnd("Februari")
Li.AddInEnd("Maret")
print("Sebelum dilakukan remove")
Li.listprint()
print("========================================")

# remove last node"
Li.RemoveEnd()
print("setela dilakukan remove node terakhir")
Li.listprint()
```

double linked list

```python
# Double Linked List
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None

class DoubleLinkedList:
    def __init__(self):
        self.head = None

    def print_list(self):
        temp = self.head
        while temp:
            print(temp.data)
            temp = temp.next

# Sisip tengah (setelah node kedua)
    def insert_tengah(self, newdata):
        new_node = Node(newdata)
        temp = self.head.next   # langsung ke node kedua

        new_node.next = temp.next
        new_node.prev = temp
        temp.next.prev = new_node
        temp.next = new_node

# Hapus dengan kata kunci
    def hapus(self, key):
        temp = self.head
        while temp:
            if temp.data == key:
                if temp.prev:
                    temp.prev.next = temp.next
                if temp.next:
                    temp.next.prev = temp.prev
                break
            temp = temp.next

# Membuat 3 node
dll = DoubleLinkedList()
dll.head = Node("A")
node2 = Node("B")
node3 = Node("C")

# Menghubungkan node
dll.head.next = node2
node2.prev = dll.head
node2.next = node3
node3.prev = node2

print("Sebelum penyisipan:")
dll.print_list()

dll.insert_tengah("D")

print("\nSetelah penyisipan tengah:")
dll.print_list()

dll.hapus("B")

print("\nSetelah hapus B:")
dll.print_list()
```

```python
# Circular Linked List

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CircularLinkedList:
    def __init__(self):
        self.head = None

    def print_list(self):
        temp = self.head
        while True:
            print(temp.data)
            temp = temp.next
            if temp == self.head:
                break

# Sisip akhir
    def insert_akhir(self, newdata):
        new_node = Node(newdata)
        temp = self.head
        while temp.next != self.head:
            temp = temp.next
        temp.next = new_node
        new_node.next = self.head

# Hapus akhir
    def hapus_akhir(self):
        temp = self.head
        prev = None
        while temp.next != self.head:
            prev = temp
            temp = temp.next
        prev.next = self.head

# Membuat 3 node
cll = CircularLinkedList()
cll.head = Node("X")
node2 = Node("Y")
node3 = Node("Z")

# Menghubungkan circular
cll.head.next = node2
node2.next = node3
node3.next = cll.head

print("Sebelum penyisipan:")
cll.print_list()

cll.insert_akhir("W")

print("\nSetelah penyisipan akhir:")
cll.print_list()

cll.hapus_akhir()

print("\nSetelah hapus akhir:")
cll.print_list()
```
